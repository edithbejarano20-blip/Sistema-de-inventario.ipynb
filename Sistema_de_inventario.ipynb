```python
class Producto:
    def __init__(self, codigo, nombre, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio

    def __str__(self):
        return f"{self.codigo} - {self.nombre} - ${self.precio}"


class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.sig = None


class Lista:
    def __init__(self):
        self.cabeza = None

    def insertar(self, producto):
        nuevo = Nodo(producto)

        if self.cabeza is None:
            self.cabeza = nuevo
            return

        actual = self.cabeza
        while actual.sig:
            actual = actual.sig

        actual.sig = nuevo

    def buscar(self, codigo):
        actual = self.cabeza

        while actual:
            if actual.dato.codigo == codigo:
                return actual.dato
            actual = actual.sig

        return None

    def eliminar(self, codigo):
        actual = self.cabeza
        anterior = None

        while actual:
            if actual.dato.codigo == codigo:
                if anterior:
                    anterior.sig = actual.sig
                else:
                    self.cabeza = actual.sig
                return True

            anterior = actual
            actual = actual.sig

        return False

    def mostrar(self):
        actual = self.cabeza
        while actual:
            print(actual.dato)
            actual = actual.sig


# COLA CON LISTA

class ColaLista:
    def __init__(self):
        self.cabeza = None
        self.final = None

    def encolar(self, dato):
        nuevo = Nodo(dato)

        if self.final is None:
            self.cabeza = self.final = nuevo
        else:
            self.final.sig = nuevo
            self.final = nuevo

    def desencolar(self):
        if self.cabeza is None:
            return None

        dato = self.cabeza.dato
        self.cabeza = self.cabeza.sig

        if self.cabeza is None:
            self.final = None

        return dato

    def vacia(self):
        return self.cabeza is None


# COLA CON ARREGLO

class ColaArreglo:
    def __init__(self):
        self.datos = []

    def encolar(self, dato):
        self.datos.append(dato)

    def desencolar(self):
        if not self.datos:
            return None
        return self.datos.pop(0)

    def vacia(self):
        return len(self.datos) == 0


# ARBOL BINARIO

class NodoArbol:
    def __init__(self, producto):
        self.producto = producto
        self.izq = None
        self.der = None


class Arbol:
    def __init__(self):
        self.raiz = None

    def insertar(self, producto):
        self.raiz = self._insertar(self.raiz, producto)

    def _insertar(self, nodo, producto):
        if nodo is None:
            return NodoArbol(producto)

        if producto.codigo < nodo.producto.codigo:
            nodo.izq = self._insertar(nodo.izq, producto)
        elif producto.codigo > nodo.producto.codigo:
            nodo.der = self._insertar(nodo.der, producto)

        return nodo

    def buscar(self, codigo):
        nodo = self.raiz

        while nodo:
            if codigo == nodo.producto.codigo:
                return nodo.producto
            elif codigo < nodo.producto.codigo:
                nodo = nodo.izq
            else:
                nodo = nodo.der

        return None

    def inorden(self, nodo):
        if nodo:
            self.inorden(nodo.izq)
            print(nodo.producto)
            self.inorden(nodo.der)


# GRAFO

class Grafo:
    def __init__(self):
        self.grafo = {}

    def conectar(self, a, b):
        self.grafo.setdefault(a, []).append(b)
        self.grafo.setdefault(b, []).append(a)

    def vecinos(self, producto):
        return self.grafo.get(producto, [])


# PRUEBA

p1 = Producto(10, "Celular", 900000)
p2 = Producto(20, "Audifonos", 80000)
p3 = Producto(30, "Funda", 50000)

lista = Lista()
lista.insertar(p1)
lista.insertar(p2)
lista.insertar(p3)

print("PRODUCTOS:")
lista.mostrar()

print("\nBUSQUEDA:")
print(lista.buscar(20))

cola = ColaLista()
cola.encolar("Pedido 1")
cola.encolar("Pedido 2")

print("\nPEDIDOS:")
print(cola.desencolar())

arbol = Arbol()
arbol.insertar(p1)
arbol.insertar(p2)
arbol.insertar(p3)

print("\nARBOL INORDEN:")
arbol.inorden(arbol.raiz)

grafo = Grafo()
grafo.conectar(10, 20)
grafo.conectar(10, 30)

print("\nPRODUCTOS RELACIONADOS:")
print(grafo.vecinos(10))
```
